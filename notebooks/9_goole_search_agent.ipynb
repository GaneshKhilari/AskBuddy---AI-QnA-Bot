```python
from dotenv import load_dotenv
load_dotenv(dotenv_path="C:\\GenAI\\.env")
```

```python
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_groq import ChatGroq
from langchain.agents import create_agent
from langgraph.checkpoint.memory import MemorySaver
```

```python
llm = ChatGroq(model ="openai/gpt-oss-20b")
```

```python
search = GoogleSerperAPIWrapper()
```

```python
search.run("please tell me which mobile phone is launched in india in last we")
```

```python
agent = create_agent(
    model =llm,
    tools =[search.run],
    system_prompt="Your are a agent and can search for any question on goole.", 
    checkpointer = MemorySaver()
)
```

```python
question ="Is the third part coming "
```

```python
response = agent.invoke(
    {"messages": [{"role": "user", "content": question}]},
    {"configurable": {"thread_id": "1"}}
    )
```

```python
print(response["messages"][-1].content)
```
