```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model="gpt-5", streaming=True)
```

```python
question = "Explain me the concept of GenAI in brief?"
```

```python
from langchain_groq import ChatGroq
llm = ChatGroq(model="openai/gpt-oss-20b",streaming=True)
```

```python
question1 = "Explain me the concept of GenAI in brief?"
```

```python
respons1 = llm.stream(question1)
for chunks in respons1:
    print(chunks.content, end="")
```
