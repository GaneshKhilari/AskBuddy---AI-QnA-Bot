## solution is: Structured output

```python
from pydantic import BaseModel, Field
from typing import List
```

```python
class ResponseStrecture(BaseModel):
    name:str = Field(description = "Complete Name")
    email:str = Field(description = "Email Address")
    age:int = Field(description = "Age")

structured_llm = llm.with_structured_output(ResponseStrecture)
```

```python
text = "Hello I am Ganesh Khilari. I am 23 years old and my emil is khilariganesh100@gmail.com"
```

```python
res = structured_llm.invoke(f"Please give me only name age and emil from given text:{text}")
res.model_dump()
```

```python
print(res.name)
print(res.age)
print(res.email)
```

```python
class Movies(BaseModel):
    title:str =Field(description="Movie Name: ")
    year: int = Field(description="Movie Released Year: ")

movie_llm = llm.with_structured_output(Movies)
```

```python
res = movie_llm.invoke("Give me one Sci-Fi movie")
res
```
