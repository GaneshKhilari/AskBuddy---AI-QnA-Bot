```python
from langchain_groq import ChatGroq
llm = ChatGroq(model="openai/gpt-oss-20b")
```

```python
res = llm.invoke("Hi I am ganesh")
print(res.content)
```

```python
res1 = llm.invoke("what is GenAI Explain me?") #Response time is very very fast as compare to other model.
print(res1.content)
```
