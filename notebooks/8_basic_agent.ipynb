```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.agents import create_agent
from langchain_core.tools import tool
```

```python
@tool
def add(a:int, b:int):
    """
    It will returns the sum of two numbers

    Args:
    a: Number one
    b: Number two

    """
    return a+b
```

```python
add.invoke({'a':5,'b':12})
```

```python
from langchain_openai import ChatOpenAI
llm= ChatOpenAI(model="gpt-5")
```

```python
from langchain.agents import create_agent
agent = create_agent(
    model=llm,
    tools=[add],
    system_prompt=""
    

)
```
