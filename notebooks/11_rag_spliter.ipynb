```python
from langchain_community.document_loaders import TextLoader
loader = TextLoader(r"..\\data\\speech.txt", encoding="utf-8")
texts = loader.load()
```

```python
from langchain_community.document_loaders import PyPDFLoader
loader = PyPDFLoader(r"..\\data\\medical_report.pdf")
pdfs = loader.load()
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=100)
```

```python
texts_chunks = splitter.split_documents(texts)
len(texts_chunks)
```

```python
pdfs_chunks = splitter.split_documents(pdfs)
len(pdfs_chunks)
```
